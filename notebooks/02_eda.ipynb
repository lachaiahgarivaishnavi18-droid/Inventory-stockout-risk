# 02 EDA

This notebook performs basic exploratory analysis, including category distribution, sales patterns, stock ranges, and stockout proportions.


```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv("data/raw/inventory_stockout_raw.csv")
print(df.head())
print(df.describe())

sns.histplot(df["current_stock"], kde=True)
plt.title("Current Stock Distribution")
plt.show()

sns.barplot(data=df, x="category", y="daily_sales", estimator="mean")
plt.title("Average Daily Sales by Category")
plt.xticks(rotation=45)
plt.show()
```
